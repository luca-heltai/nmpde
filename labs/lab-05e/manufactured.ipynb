```python
from sympy import *
x, y = symbols('x y')

prefunction = x**2*(x-1)**2*y**2*(y-1)**2

p = sin(2*pi*x)*cos(2*pi*y)

ux = -diff(prefunction, y)
uy = diff(prefunction, x)

fx = -diff(ux, x, 2) - diff(ux, y,2) - diff(p, x)
fy = -diff(uy, x, 2) - diff(uy, y,2) - diff(p, y)

fx = fx.simplify()
fy = fy.simplify()

def todealii(a):
    return str(a).replace('**', '^')
```

```python
u = todealii(str(ux)+"; " + str(uy) + "; " +str(p))
f = todealii(str(fx)+"; " + str(fy) + "; 0")
```

```python
print("set Exact solution expression =", u)
print("set Right hand side expression =", f)
```
